```python
# we cake category wise mod .
category_modes = df.groupby("Category")["Subcategory"].agg(
    lambda x: x.mode().iloc[0] if not x.mode().empty else "Unknown"
)
```

```python
# now we fill the 
df["Subcategory"] = df["Subcategory"].fillna(
    df["Category"].map(category_modes)
)
```

```python
# Calculate Q1 (25th percentile).
# This tells us the value below which 25% of transactions fall.
Q1 = df["Amount"].quantile(0.25)

# Calculate Q3 (75th percentile).
# This tells us the value below which 75% of transactions fall.
Q3 = df["Amount"].quantile(0.75)

# Calculate the Interquartile Range (IQR).
# IQR represents the middle 50% of our transaction amounts.
IQR = Q3 - Q1

# Calculate the lower statistical boundary.
# Values below this boundary are considered unusually low by the IQR method.
lower_bound = Q1 - 1.5 * IQR

# Calculate the upper statistical boundary.
# Values above this boundary are considered statistical outliers.
upper_bound = Q3 + 1.5 * IQR

# Display the values so we can understand the distribution of Amount.
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
# Select transactions that fall outside the IQR boundaries.
# These are statistical outliers, but they are NOT automatically errors.
outliers = df[
    (df["Amount"] < lower_bound) |
    (df["Amount"] > upper_bound)
]

# Count how many transactions were identified as statistical outliers.
print("Number of statistical outliers:", len(outliers))
```

```python
# Count the categories among the statistical outliers.
# This helps us understand whether large amounts are legitimate
# financial activities such as investments, salary, deposits, etc.
outliers["Category"].value_counts()
```

```python
# Count the transaction types among the outliers.
# This helps us determine whether the large amounts are mainly
# expenses, income, or money transfers.
outliers["Income/Expense"].value_counts()
```

```python
# Display the largest transactions along with useful information.
# We use this to manually verify whether the high amounts
# represent genuine financial transactions or possible data errors.
outliers[
    ["Date", "Category", "Subcategory", "Amount", "Income/Expense", "Note"]
].sort_values("Amount", ascending=False).head(20)
```
